```python

import requests
import pandas as pd

url = "https://www.ebi.ac.uk/chembl/api/data/activity.json"

params = {
    "target_chembl_id": "CHEMBL203",
    "standard_type": "IC50",
    "standard_units": "nM",
    "standard_relation": "=",
    "limit": 1000,
    "offset": 0
}

response = requests.get(url, params=params, timeout=60)
response.raise_for_status()

data = response.json()
df = pd.DataFrame(data["activities"])

print("Dataset downloaded successfully!")
print("Number of records:", len(df))
print("Columns:", df.columns.tolist())

df.head()
```

```python

print("Dataset shape:", df.shape)

columns_to_check = [
    "molecule_chembl_id",
    "standard_value",
    "standard_units",
    "canonical_smiles",
    "pchembl_value"
]

for col in columns_to_check:
    if col in df.columns:
        print(col, ":", df[col].notna().sum(), "non-missing values")
    else:
        print(col, ": not present in this response")

```

```python
print("Missing values:")
print(df.isnull().sum())

# Temporarily convert unhashable columns (lists and dicts) to string to calculate duplicate rows
df_hashable = df.copy()
for col in df_hashable.columns:
    # Check if the column contains list or dict objects
    if df_hashable[col].apply(lambda x: isinstance(x, (list, dict))).any():
        df_hashable[col] = df_hashable[col].astype(str)

print("\nDuplicate rows:", df_hashable.duplicated().sum())

print("\nDuplicate compound IDs:",
      df["molecule_chembl_id"].duplicated().sum()
      if "molecule_chembl_id" in df.columns
      else "Column not found")
```

```python

clean_df = df.copy()

clean_df["standard_value"] = pd.to_numeric(
    clean_df["standard_value"],
    errors="coerce"
)

clean_df = clean_df.dropna(
    subset=["molecule_chembl_id", "standard_value"]
)

clean_df = clean_df[
    (clean_df["standard_value"] > 0) &
    (clean_df["standard_units"] == "nM") &
    (clean_df["standard_type"] == "IC50") &
    (clean_df["standard_relation"] == "=")
].copy()

print("Original records:", len(df))
print("Records after cleaning:", len(clean_df))
print("Records removed:", len(df) - len(clean_df))

display(clean_df.head())
```

```python

import pandas as pd
import numpy as np
from rdkit import Chem

clean_df = clean_df.copy()

clean_df["canonical_smiles"] = (
    clean_df["canonical_smiles"].astype("string").str.strip()
)

molecules = [
    Chem.MolFromSmiles(s) if pd.notna(s) and s else None
    for s in clean_df["canonical_smiles"]
]

valid_mask = np.array([mol is not None for mol in molecules])

print("Total records:", len(clean_df))
print("Valid structures:", valid_mask.sum())
print("Invalid structures:", (~valid_mask).sum())

valid_df = clean_df.loc[valid_mask].copy()
valid_molecules = [
    mol for mol in molecules if mol is not None
]
```

```python

from rdkit.Chem import rdFingerprintGenerator

fingerprint_generator = rdFingerprintGenerator.GetMorganGenerator(
    radius=2,
    fpSize=2048
)

fingerprints = [
    fingerprint_generator.GetFingerprint(mol)
    for mol in valid_molecules
]

X = np.zeros((len(fingerprints), 2048), dtype=np.uint8)

from rdkit import DataStructs

for i, fp in enumerate(fingerprints):
    DataStructs.ConvertToNumpyArray(fp, X[i])

print("Fingerprint matrix shape:", X.shape)
print("First molecule's first 30 bits:")
print(X[0, :30])
```

```python

fingerprint_columns = [
    f"fp_{i}" for i in range(X.shape[1])
]

fingerprint_df = pd.DataFrame(
    X,
    columns=fingerprint_columns,
    index=valid_df.index
)

metadata_columns = [
    col for col in [
        "molecule_chembl_id",
        "canonical_smiles",
        "standard_value",
        "standard_units",
        "standard_type",
        "standard_relation"
    ]
    if col in valid_df.columns
]

final_df = pd.concat(
    [
        valid_df[metadata_columns].reset_index(drop=True),
        fingerprint_df.reset_index(drop=True)
    ],
    axis=1
)

final_df.to_csv("egfr_fingerprints.csv", index=False)

print("Saved successfully!")
print("Final dataset shape:", final_df.shape)

display(final_df.iloc[:5, :10])
```

```python

import pandas as pd
import numpy as np
from rdkit import Chem
from rdkit.Chem import rdFingerprintGenerator
from rdkit import DataStructs
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

# Prepare activity measurements
work_df = clean_df.copy()

work_df["standard_value"] = pd.to_numeric(
    work_df["standard_value"], errors="coerce"
)

work_df = work_df[
    work_df["standard_value"].notna()
    & (work_df["standard_value"] > 0)
    & work_df["molecule_chembl_id"].notna()
    & work_df["canonical_smiles"].notna()
].copy()

# Create activity labels
work_df["activity_label"] = (
    work_df["standard_value"] <= 1000
).astype(int)

# Check for conflicting labels for the same compound
label_counts = work_df.groupby(
    "molecule_chembl_id"
)["activity_label"].nunique()

conflicting_ids = label_counts[label_counts > 1].index

print("Records:", len(work_df))
print("Unique compounds:", work_df["molecule_chembl_id"].nunique())
print("Compounds with conflicting labels:", len(conflicting_ids))
```

```python

# Remove compounds with conflicting activity labels
model_df = work_df[
    ~work_df["molecule_chembl_id"].isin(conflicting_ids)
].copy()

# Keep one record per compound
model_df = model_df.drop_duplicates(
    subset="molecule_chembl_id"
).reset_index(drop=True)

print("Final unique compounds:", len(model_df))
print("\nActivity classes:")
print(model_df["activity_label"].value_counts())
```

```python

# Remove compounds with conflicting activity labels
model_df = work_df[
    ~work_df["molecule_chembl_id"].isin(conflicting_ids)
].copy()

# Keep one record per compound
model_df = model_df.drop_duplicates(
    subset="molecule_chembl_id"
).reset_index(drop=True)

print("Final unique compounds:", len(model_df))
print("\nActivity class counts:")
print(model_df["activity_label"].value_counts())

print("\nActivity class percentages:")
print(
    (model_df["activity_label"].value_counts(normalize=True) * 100).round(2)
)
```

```python

generator = rdFingerprintGenerator.GetMorganGenerator(
    radius=2,
    fpSize=2048
)

molecules = [
    Chem.MolFromSmiles(smiles)
    for smiles in model_df["canonical_smiles"]
]

valid_mask = np.array([mol is not None for mol in molecules])

model_df = model_df.loc[valid_mask].reset_index(drop=True)
molecules = [mol for mol in molecules if mol is not None]

fingerprints = [generator.GetFingerprint(mol) for mol in molecules]

X = np.zeros((len(fingerprints), 2048), dtype=np.uint8)

for i, fp in enumerate(fingerprints):
    DataStructs.ConvertToNumpyArray(fp, X[i])

y = model_df["activity_label"].to_numpy()

print("Feature matrix X:", X.shape)
print("Label vector y:", y.shape)
```

```python

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

model = RandomForestClassifier(
    n_estimators=200,
    random_state=42,
    class_weight="balanced"
)

model.fit(X_train, y_train)

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))
print("Model training completed!")
```

```python

y_pred = model.predict(X_test)

print("Accuracy:", round(accuracy_score(y_test, y_pred) * 100, 2), "%")

print("\nClassification Report:")
print(classification_report(
    y_test,
    y_pred,
    target_names=["Inactive", "Active"],
    zero_division=0
))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred))
```

```python

from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

model = RandomForestClassifier(
    n_estimators=200,
    random_state=42,
    class_weight="balanced"
)

model.fit(X_train, y_train)

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))
print("Training completed successfully!")
```

```python

from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)

y_pred = model.predict(X_test)

print("Accuracy:", round(accuracy_score(y_test, y_pred) * 100, 2), "%")

print("\nClassification Report:")
print(classification_report(
    y_test,
    y_pred,
    target_names=["Inactive", "Active"],
    zero_division=0
))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred))
```

```python

from rdkit.Chem.Scaffolds import MurckoScaffold
from sklearn.model_selection import GroupShuffleSplit
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report
import numpy as np

# Group molecules by their chemical scaffold
scaffolds = [
    MurckoScaffold.MurckoScaffoldSmiles(mol=mol)
    for mol in molecules
]

# Keep compounds with the same scaffold in the same split
splitter = GroupShuffleSplit(
    n_splits=1,
    test_size=0.20,
    random_state=42
)

train_idx, test_idx = next(
    splitter.split(X, y, groups=scaffolds)
)

X_train_sc, X_test_sc = X[train_idx], X[test_idx]
y_train_sc, y_test_sc = y[train_idx], y[test_idx]

print("Training samples:", len(train_idx))
print("Testing samples:", len(test_idx))
print("Training classes:", np.unique(y_train_sc, return_counts=True))
print("Testing classes:", np.unique(y_test_sc, return_counts=True))

# Train a fresh model
scaffold_model = RandomForestClassifier(
    n_estimators=200,
    random_state=42,
    class_weight="balanced"
)

scaffold_model.fit(X_train_sc, y_train_sc)
scaffold_pred = scaffold_model.predict(X_test_sc)

print("\nScaffold-split accuracy:",
      round(accuracy_score(y_test_sc, scaffold_pred) * 100, 2), "%")

print("\nClassification report:")
print(classification_report(
    y_test_sc,
    scaffold_pred,
    target_names=["Inactive", "Active"],
    zero_division=0
))
```

```python

from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    roc_auc_score
)

# Predicted classes
scaffold_pred = scaffold_model.predict(X_test_sc)

# Predicted probability of being active
active_prob = scaffold_model.predict_proba(X_test_sc)[:, 1]

print("Classification Report:")
print(classification_report(
    y_test_sc,
    scaffold_pred,
    target_names=["Inactive", "Active"],
    zero_division=0
))

print("Confusion Matrix:")
print(confusion_matrix(y_test_sc, scaffold_pred))

if len(set(y_test_sc)) == 2:
    print("ROC-AUC:",
          round(roc_auc_score(y_test_sc, active_prob), 3))
else:
    print("ROC-AUC unavailable: test set contains only one class.")
```
